## 3. 数値と計算

````python
print(10 + 3)   # 足し算 → 13
print(10 - 3)   # 引き算 → 7
print(10 * 3)   # 掛け算 → 30
print(10 / 3)   # 割り算 → 3.3333...（結果は常に float）
print(10 // 3)  # 割り算の商（切り捨て）→ 3
print(10 % 3)   # 余り → 1
print(10 ** 3)  # べき乗 → 1000
````

````python
# 計算の優先順位は算数と同じ（カッコ → べき乗 → 掛け算・割り算 → 足し算・引き算）
print(2 + 3 * 4)    # 14
print((2 + 3) * 4)  # 20
````

````python
# 複合代入演算子: 計算と代入をまとめて書ける
n = 10
n += 5   # n = n + 5 と同じ
print(n)  # 15
n -= 3   # n = n - 3
print(n)  # 12
n *= 2   # n = n * 2
print(n)  # 24
n //= 4  # n = n // 4
print(n)  # 6
````

````python
# 型変換
print(int("42") + 1)     # 43（文字列 → 整数）
print(float("3.14") * 2) # 6.28（文字列 → 小数）
print(str(42) + "1")     # "421"（整数 → 文字列。+ は文字列の連結になる）
print(int(3.9))          # 3（小数 → 整数。切り捨て）
print(round(3.14159, 2)) # 3.14（小数点以下 2 桁に丸める）
print(abs(-5))           # 5（絶対値）
````

````python
# 組み込み関数と math モジュール
import math

print(max(1, 9, 3))     # 最大値 → 9
print(min(1, 9, 3))     # 最小値 → 1
print(sum([1, 2, 3]))   # 合計 → 6
print(math.floor(3.7))  # 切り捨て → 3
print(math.ceil(3.2))   # 切り上げ → 4
print(math.sqrt(16))    # 平方根 → 4.0
print(math.pi)          # 円周率
````

> **注意**: コンピュータは小数を正確に表せないことがあります。`0.1 + 0.2` は `0.3` ちょうどにはなりません。お金の計算などでは注意が必要です。

````python
print(0.1 + 0.2)            # 0.30000000000000004
print(0.1 + 0.2 == 0.3)     # False
print(round(0.1 + 0.2, 2))  # 0.3 — 表示用に丸めれば OK
````

### f 文字列（f-string）

文字列の前に `f` を付けると、`{}` の中に変数や式を埋め込めます。

````python
my_name = "花子"
my_age = 25

print(f"私の名前は {my_name} です。年齢は {my_age} 歳です。")
print(f"来年は {my_age + 1} 歳になります。")

# 書式指定: 小数点以下の桁数や 3 桁区切りなど
pi = 3.14159
price = 1234567
print(f"円周率はおよそ {pi:.2f}")   # 3.14
print(f"価格: {price:,} 円")        # 1,234,567 円
````

### インデックスとスライス

文字列の各文字には位置（インデックス）があり、**0 から** 数えます。負の数を使うと末尾から数えられます。
`[開始:終了]` で一部を取り出せます（終了の位置は含まれません）。

````python
text = "Hello, Python"

print(text[0])     # 最初の文字 → "H"
print(text[-1])    # 最後の文字 → "n"
print(text[0:5])   # 0 番目から 5 番目の手前まで → "Hello"
print(text[7:])    # 7 番目から最後まで → "Python"
print(text[:5])    # 最初から 5 番目の手前まで → "Hello"
print(text[::-1])  # 逆順 → "nohtyP ,olleH"
````

````python
# 文字列の便利なメソッド
text = "Hello, Python"

print(text.upper())              # 大文字に → "HELLO, PYTHON"
print(text.lower())              # 小文字に → "hello, python"
print(text.replace("Hello", "Hi"))  # 置き換え → "Hi, Python"
print(text.find("Python"))       # 見つかった位置 → 7（なければ -1）
print(text.startswith("Hello"))  # True
print(text.endswith("Python"))   # True
print(text.count("o"))           # "o" の個数 → 2
print("Python" in text)          # 含まれているか → True
````

````python
csv = "りんご,バナナ,みかん"
fruits = csv.split(",")        # 区切り文字で分割してリストにする
print(fruits)
print(" / ".join(fruits))      # リストを文字列に結合

padded = "   余白あり   "
print(f"[{padded.strip()}]")   # 前後の空白を削除

# 複数行の文字列
poem = """一行目
二行目
三行目"""
print(poem)

# エスケープシーケンス: \n は改行、\t はタブ
print("1 行目\n2 行目")
print("名前\t年齢")
````

> **ポイント**: 文字列は変更できません（イミュータブル）。メソッドは元の文字列を変えずに、**新しい文字列を返します**。

````python
original = "hello"
upper = original.upper()
print(original)  # "hello" のまま
print(upper)     # "HELLO"
````

## 5. 比較演算子と論理演算子

| 演算子 | 意味 |
|---|---|
| `==` | 等しい |
| `!=` | 等しくない |
| `<`, `>` | より小さい、より大きい |
| `<=`, `>=` | 以下、以上 |

比較の結果は `True` か `False`（真偽値）になります。

````python
print(5 == 5)          # True
print(5 == 3)          # False
print(5 != 3)          # True
print(5 > 3)           # True
print(5 < 3)           # False
print(5 >= 5)          # True
print("abc" == "abc")  # True（文字列も比較できる）
print(1 < 2 < 3)       # True（比較をつなげて書ける）
````

### 論理演算子

| 演算子 | 意味 | 例 |
|---|---|---|
| `and` | かつ | `a and b` → 両方 True なら True |
| `or` | または | `a or b` → どちらかが True なら True |
| `not` | でない | `not a` → True と False を反転 |

````python
person_age = 20
has_ticket = True

print(person_age >= 18 and has_ticket)  # 両方 True → True
print(person_age >= 18 or has_ticket)   # どちらか True → True
print(not has_ticket)                   # 反転 → False
print(person_age < 18 and has_ticket)   # 片方 False → False
````

### 真として扱われる値・偽として扱われる値

Python では、真偽値以外の値も条件式に使えます。

**偽として扱われる値**: `False`, `0`, `0.0`, `""`（空文字列）, `None`, `[]`（空リスト）, `{}`（空辞書）— それ以外はすべて真として扱われます。

````python
print(bool(0))        # False
print(bool(""))       # False
print(bool(None))     # False
print(bool([]))       # False
print(bool(42))       # True
print(bool("hello"))  # True
print(bool([0]))      # True（要素があれば True）
````

## 6. 条件分岐

`if` 文を使うと、条件によって処理を切り替えられます。

Python では **インデント（字下げ）** でブロックを表します。条件の行末の `:` を忘れないようにしましょう。インデントには半角スペース 4 つを使うのが一般的です。

````python
temperature = 28

if temperature >= 30:
    print("暑いですね")
elif temperature >= 20:
    print("過ごしやすいですね")
else:
    print("寒いですね")
````

````python
# point の値を変えて実行してみましょう
point = 85

if point >= 80:
    print("評価: A")
elif point >= 60:
    print("評価: B")
else:
    print("評価: C")
````

### 条件式（1 行で書く if / else）

`真のときの値 if 条件 else 偽のときの値` の形で書けます。

````python
visitor_age = 15
ticket_type = "大人" if visitor_age >= 18 else "子ども"
print(ticket_type)
````

````python
# in を使った判定
day_of_week = "土"

if day_of_week in ["土", "日"]:
    print("週末です")
elif day_of_week == "月":
    print("週の始まりです")
else:
    print("平日です")
````

### match 文（Python 3.10 以降）

1 つの値を複数の候補と比較するときに使えます。`case _:` はどれにも当てはまらなかった場合です。

````python
command = "start"

match command:
    case "start":
        print("開始します")
    case "stop":
        print("停止します")
    case _:
        print("不明なコマンドです")
````

## 8. 関数

関数は「処理のまとまりに名前を付けたもの」です。`def` で定義します。同じ処理を何度も書かずに済み、プログラムが読みやすくなります。

````python
# 関数の定義
def greet(person_name):
    print(f"こんにちは、{person_name}さん！")

# 関数の呼び出し
greet("太郎")
greet("花子")
````

````python
# return で値を返す
def add(a, b):
    return a + b

result = add(3, 4)
print(result)           # 7
print(add(10, 20) * 2)  # 60（戻り値をそのまま計算に使える）
````

````python
# 引数のデフォルト値
def greet_with_title(person_name, title="さん"):
    return f"{person_name}{title}、こんにちは"

print(greet_with_title("田中"))
print(greet_with_title("田中", "先生"))

# キーワード引数: 引数名を指定して渡せる（順番を気にしなくてよい）
print(greet_with_title(title="様", person_name="佐藤"))
````

````python
# 複数の値を返す（タプルとして返される）
def min_max(values):
    return min(values), max(values)

lowest, highest = min_max([3, 1, 4, 1, 5, 9])
print(lowest, highest)
````

````python
# docstring: 関数の説明を書く（関数の最初の行に文字列を置く）
def area_of_circle(radius):
    """半径 radius の円の面積を返す"""
    return math.pi * radius ** 2

print(area_of_circle(2))
print(area_of_circle.__doc__)
````

````python
# lambda（無名関数）: 短い関数を 1 行で書く
double = lambda x: x * 2
print(double(4))  # 8

# 関数を別の関数に渡す例: sorted の key に「並べ替えの基準」を渡す
words = ["banana", "apple", "cherry"]
print(sorted(words, key=len))               # 文字数の順
print(sorted(words, key=lambda w: w[-1]))   # 最後の文字の順
````

### スコープ

関数の中で作った変数は、その関数の中でしか使えません（ローカル変数）。関数の外の変数は中から読み取れます。

````python
outside = "外側"

def show_scope():
    local = "関数の中"
    print(local, outside)  # 外側の変数は中から見える

show_scope()
# print(local) ← ここでは local は見えないのでエラー（NameError）になります
````

## 9. リスト

リストは「複数の値を順番に並べて入れるもの」です。位置（インデックス）は **0 から** 数えます。

````python
animals = ["犬", "猫", "うさぎ"]

print(animals)
print(len(animals))  # 要素数 → 3
print(animals[0])    # 最初の要素 → "犬"
print(animals[2])    # → "うさぎ"
print(animals[-1])   # 最後の要素 → "うさぎ"

animals[1] = "ハムスター"  # 要素の書き換え
print(animals)
````

````python
# 要素の追加と削除
queue = ["A", "B"]

queue.append("C")      # 末尾に追加
print(queue)           # ['A', 'B', 'C']

queue.insert(0, "Z")   # 指定した位置に追加
print(queue)           # ['Z', 'A', 'B', 'C']

last = queue.pop()     # 末尾を取り出す
print(last, queue)     # C ['Z', 'A', 'B']

queue.remove("Z")      # 値を指定して削除
print(queue)           # ['A', 'B']

queue.extend(["D", "E"])  # 別のリストを末尾に連結
print(queue)
````

````python
numbers = [3, 1, 4, 1, 5, 9, 2, 6]

print(4 in numbers)           # 含まれているか → True
print(numbers.index(5))       # 位置 → 4
print(numbers.count(1))       # 1 の個数 → 2
print(numbers[1:4])           # 一部を取り出す → [1, 4, 1]
print(sorted(numbers))        # 並べ替えた新しいリスト（元は変わらない）
print(sorted(numbers, reverse=True))  # 降順
print(sum(numbers), max(numbers), min(numbers))

numbers.sort()                # リスト自体を並べ替える
print(numbers)
numbers.reverse()             # リスト自体を逆順にする
print(numbers)
````

### リスト内包表記

「リストの各要素を変換・絞り込みした新しいリスト」を 1 行で作れる、Python らしい書き方です。

````python
prices = [100, 250, 80, 500]

# 各要素を変換
with_tax = [price * 1.1 for price in prices]
print(with_tax)

# 条件に合う要素だけを残す
expensive = [price for price in prices if price >= 200]
print(expensive)

# 同じことを for 文で書くと
with_tax2 = []
for price in prices:
    with_tax2.append(price * 1.1)
print(with_tax2)
````

````python
# 二次元リスト（リストの中にリスト）
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]
print(matrix[1][2])  # 2 行目の 3 列目 → 6

for row in matrix:
    print(" ".join(str(v) for v in row))
````

> **参照のコピーに注意**: リストを別の変数に代入すると、中身がコピーされるのではなく「同じもの」を指します。

````python
list_a = [1, 2, 3]
list_b = list_a        # 同じリストを指す
list_b.append(4)
print(list_a)          # [1, 2, 3, 4] ← list_a も変わる！

list_c = list_a.copy() # コピーを作る（list(list_a) や list_a[:] でも可）
list_c.append(5)
print(list_a)          # [1, 2, 3, 4] ← 変わらない
print(list_c)          # [1, 2, 3, 4, 5]
````

## 11. 辞書

辞書は「キー: 値」のペアをまとめたものです。関連するデータをひとまとまりにして扱えます。

````python
person = {
    "name": "山田太郎",
    "age": 30,
    "email": "taro@example.com",
    "hobbies": ["読書", "映画"],
}

print(person)
print(person["name"])       # キーで値を取り出す
print(person["hobbies"][0]) # ネストした値
print(person.get("phone"))  # 存在しないキーは get() で None が返る
print(person.get("phone", "未登録"))  # デフォルト値を指定できる
````

````python
# 要素の追加・変更・削除
person["age"] = 31        # 変更
person["city"] = "東京"    # 追加
del person["email"]       # 削除
print(person)

# キーがあるかどうか
print("city" in person)   # True
print("email" in person)  # False
````

````python
book = {"title": "Python入門", "price": 2500, "pages": 320}

print(book.keys())    # キーの一覧
print(book.values())  # 値の一覧
print(book.items())   # (キー, 値) の一覧

# for 文でキーと値を取り出す
for key, value in book.items():
    print(f"{key}: {value}")
````

### リストと辞書の組み合わせ

実際のプログラムでは「辞書のリスト」をよく扱います。

````python
students = [
    {"name": "佐藤", "score": 85},
    {"name": "鈴木", "score": 62},
    {"name": "高橋", "score": 93},
]

# 80 点以上の学生の名前を取り出す
top_students = [s["name"] for s in students if s["score"] >= 80]
print(top_students)

# 平均点
average = sum(s["score"] for s in students) / len(students)
print(f"平均点: {average:.1f}")

# 点数の高い順に並べ替え
ranking = sorted(students, key=lambda s: s["score"], reverse=True)
for i, s in enumerate(ranking, start=1):
    print(f"{i}位: {s['name']} ({s['score']}点)")
````

````python
# 辞書内包表記と、単語の出現回数を数える例
squares = {n: n ** 2 for n in range(1, 6)}
print(squares)

sentence = "the cat and the dog and the bird"
counts = {}
for word in sentence.split():
    counts[word] = counts.get(word, 0) + 1
print(counts)
````

## 12. 集合（set）

集合は「重複のない値の集まり」です。重複を取り除いたり、集合同士の演算をしたりするときに使います。

````python
unique_numbers = set([1, 2, 2, 3, 3, 3])
print(unique_numbers)        # {1, 2, 3}
print(len(unique_numbers))   # 3

unique_numbers.add(4)
print(unique_numbers)
print(2 in unique_numbers)   # True

a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a | b)   # 和集合 → {1, 2, 3, 4, 5}
print(a & b)   # 積集合 → {3, 4}
print(a - b)   # 差集合 → {1, 2}
````

## 13. クラス

クラスは「オブジェクトの設計図」です。同じ構造のオブジェクトをたくさん作るときに使います。

- `__init__` はオブジェクトを作るときに自動で呼ばれる特別なメソッドです
- メソッドの最初の引数 `self` は「そのオブジェクト自身」を表します

````python
class Dog:
    def __init__(self, dog_name, breed):
        self.name = dog_name
        self.breed = breed

    def bark(self):
        print(f"{self.name}: ワン！")

    def introduce(self):
        return f"{self.name} は {self.breed} です"


pochi = Dog("ポチ", "柴犬")
hachi = Dog("ハチ", "秋田犬")

pochi.bark()
print(hachi.introduce())
print(pochi.name)
````

````python
# 継承: 既存のクラスをもとに新しいクラスを作る
class Animal:
    def __init__(self, animal_name):
        self.name = animal_name

    def speak(self):
        print(f"{self.name} が鳴きます")

    def __str__(self):
        # print() したときの表示を決める特別なメソッド
        return f"Animal({self.name})"


class Cat(Animal):
    def __init__(self, cat_name):
        super().__init__(cat_name)  # 親クラスの __init__ を呼ぶ

    def speak(self):
        # メソッドの上書き（オーバーライド）
        print(f"{self.name}: ニャー")


tama = Cat("タマ")
tama.speak()
print(tama)                      # __str__ の結果が表示される
print(isinstance(tama, Cat))     # True
print(isinstance(tama, Animal))  # True
````

````python
# クラス属性と @property
class Rectangle:
    count = 0  # クラス属性: すべてのオブジェクトで共有される

    def __init__(self, width, height):
        self.width = width    # インスタンス属性: オブジェクトごとに持つ
        self.height = height
        Rectangle.count += 1

    @property
    def area(self):
        # @property を付けると、メソッドをプロパティのように () なしで呼べる
        return self.width * self.height


rect1 = Rectangle(3, 4)
rect2 = Rectangle(5, 5)
print(rect1.area)        # 12（() は不要）
print(rect2.area)        # 25
print(Rectangle.count)   # 2（作られたオブジェクトの数）
````

## 14. 例外処理

エラー（例外）が起きてもプログラムを止めずに対処するには `try` / `except` を使います。

````python
try:
    # エラーが起きるかもしれない処理
    number = int("これは数字ではない")
    print(number)
except ValueError as error:
    # エラーが起きたときの処理
    print("エラーが発生しました:", error)
finally:
    # エラーの有無にかかわらず必ず実行される
    print("処理を終了します")
````

````python
# raise で自分でエラーを発生させる
def divide(a, b):
    if b == 0:
        raise ValueError("0 で割ることはできません")
    return a / b

try:
    print(divide(10, 2))
    print(divide(10, 0))  # ここでエラーが発生する
    print("この行は実行されません")
except ValueError as error:
    print("キャッチ:", error)
````

| エラー | 原因の例 |

| `SyntaxError` | 文法の間違い（`:` やカッコの閉じ忘れなど） |
| `IndentationError` | インデントがそろっていない |
| `NameError` | 定義していない変数を使った |
| `TypeError` | 型に合わない操作（数値と文字列を足すなど） |
| `ValueError` | 型は合っているが値が不正（`int("abc")` など） |
| `IndexError` | リストの範囲外の位置を指定した |
| `KeyError` | 辞書に存在しないキーを指定した |
| `ZeroDivisionError` | 0 で割った |

エラーメッセージをよく読むことが、デバッグの第一歩です。

````python
# エラーの種類を確認してみる（try / except で受け止めているのでノートブックは止まりません）
tests = [
    lambda: undefined_variable + 1,
    lambda: 1 + "1",
    lambda: [1, 2, 3][10],
    lambda: {"a": 1}["b"],
    lambda: 10 / 0,
]

for test in tests:
    try:
        test()
    except Exception as error:
        print(type(error).__name__, "→", error)
````

## 16. ファイル操作

`open()` でファイルを読み書きできます。`with` を使うと、ブロックを抜けたときに自動でファイルが閉じられます。

````python
# 書き込み（"w" モード: ファイルがなければ作り、あれば上書き）
with open("memo.txt", "w", encoding="utf-8") as f:
    f.write("1 行目\n")
    f.write("2 行目\n")
    f.write("3 行目\n")

# 読み込み（"r" モード）
with open("memo.txt", "r", encoding="utf-8") as f:
    content = f.read()
print(content)

# 1 行ずつ読む
with open("memo.txt", "r", encoding="utf-8") as f:
    for line_number, line in enumerate(f, start=1):
        print(line_number, line.strip())
````

````python
# 追記（"a" モード）と後片付け
import os

with open("memo.txt", "a", encoding="utf-8") as f:
    f.write("4 行目（追記）\n")

with open("memo.txt", "r", encoding="utf-8") as f:
    print(f.read())

os.remove("memo.txt")                 # ファイルを削除
print(os.path.exists("memo.txt"))     # False
````

### 問題 5: クラスを作る

銀行口座を表す `BankAccount` クラスを作ってください。

- `__init__(self, owner, balance=0)`: 口座名義と残高（初期値 0）
- `deposit(self, amount)`: 入金する
- `withdraw(self, amount)`: 出金する。残高が足りなければ `raise ValueError("残高が不足しています")`
- `show(self)`: `山田 さんの残高: 1200 円` のように表示する

````python
# 問題 5: ここにコードを書いてください
````

**解答例**

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("残高が不足しています")
        self.balance -= amount

    def show(self):
        print(f"{self.owner} さんの残高: {self.balance} 円")


account = BankAccount("山田", 1000)
account.deposit(500)
account.withdraw(300)
account.show()  # 山田 さんの残高: 1200 円

try:
    account.withdraw(5000)
except ValueError as error:
    print("エラー:", error)
````
