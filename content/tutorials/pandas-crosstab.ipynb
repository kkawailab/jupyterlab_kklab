### 解答例 2

````python
pd.crosstab(
    index=df["学年"],
    columns=df["生成AI利用頻度"]
)
````

### ワンポイント — 列の並び順を整える

上の表は列が文字コード順(時々 → 未経験 → 毎週)に並んでしまいます。
「未経験 → 時々 → 毎週」という自然な順序にするには、列を**順序付きカテゴリ型**に変換しておきます。
以降の課題の表もすべてこの順序で表示されるようになります。

````python
df["生成AI利用頻度"] = pd.Categorical(
    df["生成AI利用頻度"],
    categories=["未経験", "時々", "毎週"],
    ordered=True,
)

crosstab_grade = pd.crosstab(index=df["学年"], columns=df["生成AI利用頻度"])
crosstab_grade
````

`margins=True` を指定すると、合計の行・列(既定の名前は `All`)が追加されます。
`margins_name` で合計行・列の名前を日本語にすることもできます。

````python
# ここにコードを書いてみましょう
````

## 課題 4 — 行方向の構成比(百分率)

「学年ごとに、未経験・時々・毎週が何 % ずつか」を知りたいときは `normalize="index"` を使います。
各行の合計が 1 になるように割合が計算されるので、100 倍して百分率にし、見やすく丸めます。

````python
# ここにコードを書いてみましょう
````

### 解答例 4

````python
ratio = pd.crosstab(
    index=df["学年"],
    columns=df["生成AI利用頻度"],
    normalize="index",
)
(ratio * 100).round(1)
````

各行を横に足すと 100% になっています。
「学年が上がるほど『毎週』の割合が増えていく」ことが、人数の表よりもはっきり読み取れます。

## 発展課題 — 件数ではなく「平均値」のクロス表

`pd.crosstab()` は `values` と `aggfunc` を指定すると、件数の代わりに
**組み合わせごとの集計値**(平均など)を計算できます。
まだ使っていない「週間学習時間」の列を使って、学年 × 利用頻度ごとの平均学習時間を求めてみましょう。

````python
# ここにコードを書いてみましょう
````

### 解答例(発展課題)

````python
pd.crosstab(
    index=df["学年"],
    columns=df["生成AI利用頻度"],
    values=df["週間学習時間"],
    aggfunc="mean",
).round(1)
````

学年が上がるほど、また利用頻度が高いグループほど、平均学習時間が長い傾向が見えます
(空欄 `NaN` は「その組み合わせの学生がいない」という意味です)。

ただしこれは架空データの相関であり、「生成 AI を使うと学習時間が延びる」という**因果関係を示すものではない**ことに注意しましょう。

## まとめ

| やりたいこと | 書き方 |
|---|---|
| 基本のクロス集計 | `pd.crosstab(index=df["行"], columns=df["列"])` |
| 合計を追加 | `margins=True`(名前は `margins_name` で変更可) |
| 行方向の構成比 | `normalize="index"`(列方向は `"columns"`、全体は `"all"`) |
| 組み合わせごとの平均など | `values=df["数値列"], aggfunc="mean"` |
| 列の並び順を整える | `pd.Categorical(..., categories=[...], ordered=True)` |

クロス集計は「2 つのカテゴリの関係を最初にざっくりつかむ」ための基本の道具です。
より本格的な分析(検定など)は `statistics/` フォルダーの統計ノートブックで扱っています。
