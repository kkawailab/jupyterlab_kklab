## 2. 変数と定数

変数は「値を入れておく箱」です。JavaScript では `let` と `const` を使って宣言します。

| キーワード | 意味 | 値の変更（再代入） |
|---|---|---|
| `let` | 変数 | できる |
| `const` | 定数 | できない |
| `var` | 古い書き方の変数 | できる（現在は `let` / `const` を使うのが一般的） |

````javascript
// let: あとから値を変更できる
let score = 10;
console.log("最初の score:", score);

score = 20;         // 値を変更（再代入）
console.log("変更後の score:", score);

score = score + 5;  // 今の値に 5 を足す
console.log("さらに変更後:", score);
````

````javascript
// const: 一度決めたら変更できない（定数）
const TAX_RATE = 0.1;
const appName = "MyApp";

console.log(TAX_RATE);
console.log(appName);

// TAX_RATE = 0.2; とすると「TypeError: Assignment to constant variable.」というエラーになります
````

## 3. データ型

JavaScript の値には「型」があります。`typeof` 演算子で型を調べられます。

| 型 | 例 | 説明 |
|---|---|---|
| `number` | `42`, `3.14` | 数値（整数も小数も同じ型） |
| `string` | `"hello"` | 文字列 |
| `boolean` | `true`, `false` | 真偽値 |
| `undefined` | `undefined` | 値がまだ入っていない |
| `null` | `null` | 「何もない」ことを意図的に表す |
| `object` | `{}`, `[]` | オブジェクト・配列など |
| `function` | `function() {}` | 関数 |

````javascript
console.log(typeof 42);             // number
console.log(typeof 3.14);           // number
console.log(typeof "hello");        // string
console.log(typeof true);           // boolean
console.log(typeof undefined);      // undefined
console.log(typeof null);           // object（歴史的な理由でこうなっています）
console.log(typeof [1, 2, 3]);      // object（配列もオブジェクトの一種）
console.log(typeof { a: 1 });       // object
console.log(typeof function () {}); // function
````

````javascript
// 宣言しただけで値を入れていない変数は undefined
let nothing;
console.log(nothing);

// 「値がない」ことを明示的に表すには null を使う
let selectedItem = null;
console.log(selectedItem);
````

## 4. 数値と計算

````javascript
console.log(10 + 3);   // 足し算 → 13
console.log(10 - 3);   // 引き算 → 7
console.log(10 * 3);   // 掛け算 → 30
console.log(10 / 3);   // 割り算 → 3.3333...
console.log(10 % 3);   // 余り → 1
console.log(10 ** 3);  // べき乗 → 1000
````

````javascript
// 計算の優先順位は算数と同じ（カッコ → 掛け算・割り算 → 足し算・引き算）
console.log(2 + 3 * 4);    // 14
console.log((2 + 3) * 4);  // 20
````

````javascript
// 複合代入演算子: 計算と代入をまとめて書ける
let n = 10;
n += 5;  // n = n + 5 と同じ
console.log(n); // 15
n -= 3;  // n = n - 3
console.log(n); // 12
n *= 2;  // n = n * 2
console.log(n); // 24
n /= 4;  // n = n / 4
console.log(n); // 6

// インクリメント・デクリメント（1 増やす・1 減らす）
n++;
console.log(n); // 7
n--;
console.log(n); // 6
````

````javascript
// Math オブジェクト: 便利な数学関数
console.log(Math.floor(3.7));   // 切り捨て → 3
console.log(Math.ceil(3.2));    // 切り上げ → 4
console.log(Math.round(3.5));   // 四捨五入 → 4
console.log(Math.abs(-5));      // 絶対値 → 5
console.log(Math.max(1, 9, 3)); // 最大値 → 9
console.log(Math.min(1, 9, 3)); // 最小値 → 1
console.log(Math.sqrt(16));     // 平方根 → 4
console.log(Math.PI);           // 円周率

// 0 以上 1 未満のランダムな数（実行するたびに変わります）
console.log(Math.random());
````

````javascript
// 文字列と数値の変換
console.log(Number("42") + 1);      // 43（文字列 → 数値）
console.log(parseInt("42px"));      // 42（先頭の整数部分を取り出す）
console.log(parseFloat("3.14abc")); // 3.14
console.log(String(42) + 1);        // "421"（数値 → 文字列。+ は文字列の連結になる）
console.log((3.14159).toFixed(2));  // "3.14"（小数点以下2桁の文字列にする）
````

> **注意**: コンピュータは小数を正確に表せないことがあります。`0.1 + 0.2` は `0.3` ちょうどにはなりません。お金の計算などでは注意が必要です。

````javascript
console.log(0.1 + 0.2);              // 0.30000000000000004
console.log(0.1 + 0.2 === 0.3);      // false
console.log((0.1 + 0.2).toFixed(2)); // "0.30" — 表示用に丸めれば OK
````

## 5. 文字列

````javascript
// 文字列はシングルクォート(')、ダブルクォート(")、バッククォート(`)で囲む
let s1 = 'シングルクォート';
let s2 = "ダブルクォート";
let s3 = `バッククォート`;
console.log(s1, s2, s3);

// 文字列の連結（+）
let firstName = "太郎";
let lastName = "山田";
console.log(lastName + " " + firstName);
````

````javascript
// テンプレートリテラル: バッククォートで囲むと ${} の中に変数や式を埋め込める
let myName = "花子";
let myAge = 25;

console.log(`私の名前は ${myName} です。年齢は ${myAge} 歳です。`);
console.log(`来年は ${myAge + 1} 歳になります。`);

// 複数行の文字列も書ける
let poem = `一行目
二行目
三行目`;
console.log(poem);
````

````javascript
// 文字列の便利なプロパティ・メソッド
let text = "Hello, JavaScript";

console.log(text.length);                 // 文字数 → 17
console.log(text.toUpperCase());          // 大文字に → "HELLO, JAVASCRIPT"
console.log(text.toLowerCase());          // 小文字に → "hello, javascript"
console.log(text.includes("Java"));       // 含まれているか → true
console.log(text.indexOf("Java"));        // 見つかった位置 → 7（0 から数える）
console.log(text.slice(0, 5));            // 0 番目から 5 番目の手前まで → "Hello"
console.log(text.replace("Hello", "Hi")); // 置き換え → "Hi, JavaScript"
console.log(text[0]);                     // 0 番目の文字 → "H"
console.log(text.at(-1));                 // 最後の文字 → "t"
````

````javascript
let csv = "りんご,バナナ,みかん";
let fruits = csv.split(",");        // 区切り文字で分割して配列にする
console.log(fruits);
console.log(fruits.join(" / "));    // 配列を文字列に結合

let padded = "   余白あり   ";
console.log(`[${padded.trim()}]`);  // 前後の空白を削除

console.log("abc".repeat(3));                 // 繰り返し → "abcabcabc"
console.log("JavaScript".startsWith("Java")); // true
console.log("JavaScript".endsWith("Script")); // true
````

> **ポイント**: 文字列のメソッドは元の文字列を変更せず、**新しい文字列を返します**。

````javascript
let original = "hello";
let upper = original.toUpperCase();
console.log(original); // "hello" のまま
console.log(upper);    // "HELLO"
````

## 6. 比較演算子と論理演算子

| 演算子 | 意味 |
|---|---|
| `===` | 等しい（型も値も同じ） |
| `!==` | 等しくない |
| `<`, `>` | より小さい、より大きい |
| `<=`, `>=` | 以下、以上 |

比較の結果は `true` か `false`（真偽値）になります。

````javascript
console.log(5 === 5);         // true
console.log(5 === 3);         // false
console.log(5 !== 3);         // true
console.log(5 > 3);           // true
console.log(5 < 3);           // false
console.log(5 >= 5);          // true
console.log("abc" === "abc"); // true（文字列も比較できる）
````

> **`==` と `===` の違い**: `==` は型を自動変換してから比較するため、思わぬ結果になることがあります。常に `===` と `!==` を使いましょう。

````javascript
console.log(5 == "5");    // true （型変換されて等しいと判定される）
console.log(5 === "5");   // false（型が違うので等しくない）
console.log(0 == false);  // true
console.log(0 === false); // false
````

### 論理演算子

| 演算子 | 意味 | 例 |
|---|---|---|
| `&&` | かつ（AND） | `a && b` → 両方 true なら true |
| `\|\|` | または（OR） | `a \|\| b` → どちらかが true なら true |
| `!` | でない（NOT） | `!a` → true と false を反転 |

````javascript
let personAge = 20;
let hasTicket = true;

console.log(personAge >= 18 && hasTicket); // 両方 true → true
console.log(personAge >= 18 || hasTicket); // どちらか true → true
console.log(!hasTicket);                   // 反転 → false
console.log(personAge < 18 && hasTicket);  // 片方 false → false
````

### truthy と falsy

JavaScript では、真偽値以外の値も「真っぽい（truthy）」「偽っぽい（falsy）」として扱われます。

**falsy な値**: `false`, `0`, `""`（空文字列）, `null`, `undefined`, `NaN` — それ以外はすべて truthy です。

````javascript
console.log(Boolean(0));         // false
console.log(Boolean(""));        // false
console.log(Boolean(null));      // false
console.log(Boolean(undefined)); // false
console.log(Boolean(42));        // true
console.log(Boolean("hello"));   // true
console.log(Boolean([]));        // true（空の配列も truthy）
````

### switch 文

1 つの値を複数の候補と比較するときに便利です。各 `case` の最後に `break` を忘れずに。

````javascript
let dayOfWeek = "土";

switch (dayOfWeek) {
  case "土":
  case "日":
    console.log("週末です");
    break;
  case "月":
    console.log("週の始まりです");
    break;
  default:
    console.log("平日です");
}
````

## 8. 繰り返し（ループ）

````javascript
// 1 から 10 までの合計
let sum = 0;
for (let i = 1; i <= 10; i++) {
  sum += i;
}
console.log(`合計: ${sum}`);
````

````javascript
// for...of: 配列の要素を 1 つずつ取り出す
let colors = ["赤", "緑", "青"];
for (const color of colors) {
  console.log(color);
}

// 文字列も 1 文字ずつ取り出せる
for (const ch of "abc") {
  console.log(ch);
}
````

````javascript
// break: ループを途中で抜ける / continue: 次の繰り返しへ進む
for (let i = 1; i <= 10; i++) {
  if (i % 2 === 0) {
    continue; // 偶数はスキップ
  }
  if (i > 7) {
    break;    // 7 を超えたら終了
  }
  console.log(i);
}
// 出力: 1, 3, 5, 7
````

## 9. 関数

関数は「処理のまとまりに名前を付けたもの」です。同じ処理を何度も書かずに済み、プログラムが読みやすくなります。

````javascript
// 関数の定義
function greet(personName) {
  console.log(`こんにちは、${personName}さん！`);
}

// 関数の呼び出し
greet("太郎");
greet("花子");
````

````javascript
// return で値を返す
function add(a, b) {
  return a + b;
}

let result = add(3, 4);
console.log(result);          // 7
console.log(add(10, 20) * 2); // 60（戻り値をそのまま計算に使える）
````

````javascript
// 引数のデフォルト値
function greetWithTitle(personName, title = "さん") {
  return `${personName}${title}、こんにちは`;
}

console.log(greetWithTitle("田中"));
console.log(greetWithTitle("田中", "先生"));
````

````javascript
// アロー関数: 関数をより短く書ける
const square = (x) => {
  return x * x;
};

// 本体が 1 つの式だけなら {} と return を省略できる
const double = (x) => x * 2;

// 引数が 1 つなら () も省略できる
const triple = x => x * 3;

console.log(square(4)); // 16
console.log(double(4)); // 8
console.log(triple(4)); // 12

// 引数が複数・なしの場合
const multiply = (a, b) => a * b;
const sayHi = () => "Hi!";
console.log(multiply(3, 5));
console.log(sayHi());
````

````javascript
// 関数を別の関数に渡す（コールバック関数）
function repeatTimes(count, action) {
  for (let i = 0; i < count; i++) {
    action(i);
  }
}

repeatTimes(3, (i) => {
  console.log(`${i + 1} 回目の呼び出し`);
});
````

> 「1 つの関数は 1 つの仕事をする」を意識すると、読みやすく再利用しやすいコードになります。

## 10. 配列

配列は「複数の値を順番に並べて入れるもの」です。位置（インデックス）は **0 から** 数えます。

````javascript
let animals = ["犬", "猫", "うさぎ"];

console.log(animals);
console.log(animals.length);  // 要素数 → 3
console.log(animals[0]);      // 最初の要素 → "犬"
console.log(animals[2]);      // → "うさぎ"
console.log(animals.at(-1));  // 最後の要素 → "うさぎ"

animals[1] = "ハムスター";     // 要素の書き換え
console.log(animals);
````

````javascript
// 要素の追加と削除
let queue = ["A", "B"];

queue.push("C");     // 末尾に追加
console.log(queue);  // ["A", "B", "C"]

queue.unshift("Z");  // 先頭に追加
console.log(queue);  // ["Z", "A", "B", "C"]

let last = queue.pop();    // 末尾を取り出す
console.log(last, queue);  // "C" ["Z", "A", "B"]

let first = queue.shift(); // 先頭を取り出す
console.log(first, queue); // "Z" ["A", "B"]
````

````javascript
let numbers = [3, 1, 4, 1, 5, 9, 2, 6];

console.log(numbers.includes(4));                // 含まれているか → true
console.log(numbers.indexOf(5));                 // 位置 → 4
console.log(numbers.slice(1, 4));                // 一部を取り出す → [1, 4, 1]
console.log(numbers.join("-"));                  // 文字列に結合 → "3-1-4-1-5-9-2-6"
console.log([...numbers].reverse());             // 逆順（コピーしてから）
console.log([...numbers].sort((a, b) => a - b)); // 昇順に並べ替え
````

### 配列を処理するメソッド

配列には「各要素に対して関数を適用する」便利なメソッドがあります。まずは `forEach`, `map`, `filter` の 3 つを覚えましょう。

````javascript
let prices = [100, 250, 80, 500];

// forEach: 各要素に対して処理を行う
prices.forEach((price) => {
  console.log(`${price} 円`);
});
````

````javascript
// map: 各要素を変換した新しい配列を作る
let withTax = prices.map((price) => price * 1.1);
console.log(withTax);

// filter: 条件に合う要素だけの新しい配列を作る
let expensive = prices.filter((price) => price >= 200);
console.log(expensive);

// find: 条件に合う最初の要素を返す
let firstCheap = prices.find((price) => price < 100);
console.log(firstCheap);

// some / every: 条件に合う要素が「1 つでもあるか」/「すべてか」
console.log(prices.some((price) => price > 400));  // true
console.log(prices.every((price) => price > 50));  // true
````

````javascript
// reduce: 配列を 1 つの値にまとめる（合計など）
let totalPrice = prices.reduce((acc, price) => acc + price, 0);
console.log(`合計: ${totalPrice} 円`);

// メソッドはつなげて書ける（メソッドチェーン）
let discountedTotal = prices
  .filter((price) => price >= 100)  // 100 円以上だけ
  .map((price) => price * 0.9)      // 10% 引き
  .reduce((acc, price) => acc + price, 0);
console.log(discountedTotal);
````

````javascript
// スプレッド構文（...）: 配列を展開する
let arr1 = [1, 2, 3];
let arr2 = [4, 5];
let combined = [...arr1, ...arr2];
console.log(combined);          // [1, 2, 3, 4, 5]
console.log(Math.max(...arr1)); // 関数の引数に展開 → 3

// 分割代入: 配列の要素を個別の変数に取り出す
let [x, y, z] = [10, 20, 30];
console.log(x, y, z);
````

````javascript
// 二次元配列（配列の中に配列）
let matrix = [
  [1, 2, 3],
  [4, 5, 6],
];
console.log(matrix[1][2]); // 2 行目の 3 列目 → 6

for (const row of matrix) {
  console.log(row.join(" "));
}
````

## 11. オブジェクト

オブジェクトは「キー: 値」のペアをまとめたものです。関連するデータをひとまとまりにして扱えます。

````javascript
let person = {
  name: "山田太郎",
  age: 30,
  email: "taro@example.com",
  hobbies: ["読書", "映画"],
};

console.log(person);
console.log(person.name);       // ドット記法
console.log(person["age"]);     // ブラケット記法（キーを文字列で指定）
console.log(person.hobbies[0]); // ネストした値
````

````javascript
// プロパティの追加・変更・削除
person.age = 31;       // 変更
person.city = "東京";   // 追加
delete person.email;   // 削除
console.log(person);

// プロパティがあるかどうか
console.log("city" in person); // true
console.log(person.phone);     // 存在しないプロパティは undefined
````

````javascript
// メソッド: オブジェクトの中の関数。this でそのオブジェクト自身を参照できる
let counter = {
  value: 0,
  increment() {
    this.value++;
  },
  show() {
    console.log(`現在の値: ${this.value}`);
  },
};

counter.increment();
counter.increment();
counter.show(); // 現在の値: 2
````

````javascript
let book = { title: "JavaScript入門", price: 2500, pages: 320 };

console.log(Object.keys(book));    // キーの配列
console.log(Object.values(book));  // 値の配列
console.log(Object.entries(book)); // [キー, 値] の配列

// for...in: キーを 1 つずつ取り出す
for (const key in book) {
  console.log(`${key}: ${book[key]}`);
}
````

````javascript
// オプショナルチェーン（?.）: 途中が null / undefined でもエラーにならない
let user = { profile: { nickname: "たろ" } };
let guest = {};

console.log(user.profile?.nickname);  // "たろ"
console.log(guest.profile?.nickname); // undefined（エラーにならない）

// Null 合体演算子（??）: 左が null / undefined のときだけ右を使う
console.log(guest.profile?.nickname ?? "ゲスト"); // "ゲスト"
````

### 配列とオブジェクトの組み合わせ

実際のプログラムでは「オブジェクトの配列」をよく扱います。

````javascript
let students = [
  { name: "佐藤", score: 85 },
  { name: "鈴木", score: 62 },
  { name: "高橋", score: 93 },
];

// 80 点以上の学生の名前を取り出す
let topStudents = students
  .filter((s) => s.score >= 80)
  .map((s) => s.name);
console.log(topStudents);

// 平均点
let average = students.reduce((acc, s) => acc + s.score, 0) / students.length;
console.log(`平均点: ${average.toFixed(1)}`);

// 点数の高い順に並べ替え
let ranking = [...students].sort((a, b) => b.score - a.score);
ranking.forEach((s, i) => console.log(`${i + 1}位: ${s.name} (${s.score}点)`));
````

> **参照のコピーに注意**: オブジェクトや配列を別の変数に代入すると、中身がコピーされるのではなく「同じもの」を指します。

````javascript
let listA = [1, 2, 3];
let listB = listA;      // 同じ配列を指す
listB.push(4);
console.log(listA);     // [1, 2, 3, 4] ← listA も変わる！

let listC = [...listA]; // スプレッド構文でコピーを作る
listC.push(5);
console.log(listA);     // [1, 2, 3, 4] ← 変わらない
console.log(listC);     // [1, 2, 3, 4, 5]
````

````javascript
let outside = "外側";

if (true) {
  let inside = "内側";
  console.log(outside); // 外側の変数は中から見える
  console.log(inside);
}

// console.log(inside); ← ここでは inside は見えないのでエラー（ReferenceError）になります

function showScope() {
  let local = "関数の中";
  console.log(local, outside);
}
showScope();
// console.log(local); ← これもエラーになります
````

## 13. クラス

クラスは「オブジェクトの設計図」です。同じ構造のオブジェクトをたくさん作るときに使います。

````javascript
class Dog {
  // constructor: new でオブジェクトを作るときに呼ばれる
  constructor(dogName, breed) {
    this.name = dogName;
    this.breed = breed;
  }

  // メソッド
  bark() {
    console.log(`${this.name}: ワン！`);
  }

  introduce() {
    return `${this.name} は ${this.breed} です`;
  }
}

let pochi = new Dog("ポチ", "柴犬");
let hachi = new Dog("ハチ", "秋田犬");

pochi.bark();
console.log(hachi.introduce());
````

````javascript
// 継承: 既存のクラスをもとに新しいクラスを作る
class Animal {
  constructor(animalName) {
    this.name = animalName;
  }
  speak() {
    console.log(`${this.name} が鳴きます`);
  }
}

class Cat extends Animal {
  constructor(catName) {
    super(catName); // 親クラスの constructor を呼ぶ
  }
  speak() {
    // メソッドの上書き（オーバーライド）
    console.log(`${this.name}: ニャー`);
  }
}

let tama = new Cat("タマ");
tama.speak();
console.log(tama instanceof Cat);    // true
console.log(tama instanceof Animal); // true
````

````javascript
// getter と static メソッド
class Rectangle {
  constructor(width, height) {
    this.width = width;
    this.height = height;
  }

  get area() {          // getter: プロパティのように呼び出せる
    return this.width * this.height;
  }

  static square(size) { // static: クラスから直接呼ぶ
    return new Rectangle(size, size);
  }
}

let rect = new Rectangle(3, 4);
console.log(rect.area);                // 12（() は不要）
console.log(Rectangle.square(5).area); // 25
````

## 14. エラー処理

エラーが起きてもプログラムを止めずに対処するには `try` / `catch` を使います。

````javascript
try {
  // エラーが起きるかもしれない処理
  let data = JSON.parse("これはJSONではない");
  console.log(data);
} catch (error) {
  // エラーが起きたときの処理
  console.log("エラーが発生しました:", error.message);
} finally {
  // エラーの有無にかかわらず必ず実行される
  console.log("処理を終了します");
}
````

````javascript
// throw で自分でエラーを投げる
function divide(a, b) {
  if (b === 0) {
    throw new Error("0 で割ることはできません");
  }
  return a / b;
}

try {
  console.log(divide(10, 2));
  console.log(divide(10, 0)); // ここでエラーが投げられる
  console.log("この行は実行されません");
} catch (error) {
  console.log("キャッチ:", error.message);
}
````

### よくあるエラーの種類

| エラー | 原因の例 |
|---|---|
| `SyntaxError` | 文法の間違い（カッコの閉じ忘れなど） |
| `ReferenceError` | 宣言していない変数を使った |
| `TypeError` | 型に合わない操作（`null` のプロパティを読む、定数に再代入するなど） |
| `RangeError` | 範囲外の値（無限に再帰する関数など） |

エラーメッセージをよく読むことが、デバッグの第一歩です。

````javascript
// エラーの種類を確認してみる（try / catch で受け止めているのでノートブックは止まりません）
try {
  undefinedVariable + 1;
} catch (error) {
  console.log(error.name, "→", error.message);
}

try {
  let nothingHere = null;
  nothingHere.someProperty;
} catch (error) {
  console.log(error.name, "→", error.message);
}

try {
  const FIXED = 1;
  FIXED = 2;
} catch (error) {
  console.log(error.name, "→", error.message);
}
````

## 15. 非同期処理（入門）

JavaScript では「時間のかかる処理」（タイマー、通信など）を待っている間に、他の処理を進められます。これを **非同期処理** といいます。ここでは基本だけ紹介します。

````javascript
// setTimeout: 指定した時間（ミリ秒）後に関数を実行する
console.log("1: 開始");

setTimeout(() => {
  console.log("3: 1 秒後に実行されました");
}, 1000);

console.log("2: setTimeout の後（こちらが先に実行される）");

// このセルの実行が終わるまで少し待つ（後述の await）
await new Promise((resolve) => setTimeout(resolve, 1500));
````

````javascript
// Promise と async / await
function wait(ms) {
  return new Promise((resolve) => setTimeout(resolve, ms));
}

async function countDown() {
  for (let i = 3; i >= 1; i--) {
    console.log(i);
    await wait(500); // 0.5 秒待つ
  }
  console.log("完了!");
  return "終わりました";
}

// このノートブックでは、セルの中で直接 await が使えます
let doneMessage = await countDown();
console.log(doneMessage);
````

## 17. Map と Set

配列・オブジェクトに加えて、`Map`（キーと値のペア）と `Set`（重複のない集合）もよく使います。

````javascript
// Map: キーと値のペア（キーに何でも使える、追加した順序が保たれる）
let stock = new Map();
stock.set("りんご", 5);
stock.set("バナナ", 0);
console.log(stock.get("りんご")); // 5
console.log(stock.has("バナナ")); // true
console.log(stock.size);          // 2

for (const [item, count] of stock) {
  console.log(`${item}: ${count} 個`);
}

// Set: 重複のない値の集合
let uniqueNumbers = new Set([1, 2, 2, 3, 3, 3]);
console.log(uniqueNumbers);       // Set(3) { 1, 2, 3 }
console.log([...uniqueNumbers]);  // 配列に変換 → [1, 2, 3]
````

### 問題 5: クラスを作る

銀行口座を表す `BankAccount` クラスを作ってください。

- `constructor(owner, balance)`: 口座名義と残高（初期値 0）
- `deposit(amount)`: 入金する
- `withdraw(amount)`: 出金する。残高が足りなければ `throw new Error("残高が不足しています")`
- `show()`: `山田 さんの残高: 1200 円` のように表示する

````javascript
// 問題 5: ここにコードを書いてください
````

**解答例**

````javascript
class BankAccount {
  constructor(owner, balance = 0) {
    this.owner = owner;
    this.balance = balance;
  }

  deposit(amount) {
    this.balance += amount;
  }

  withdraw(amount) {
    if (amount > this.balance) {
      throw new Error("残高が不足しています");
    }
    this.balance -= amount;
  }

  show() {
    console.log(`${this.owner} さんの残高: ${this.balance} 円`);
  }
}

const account = new BankAccount("山田", 1000);
account.deposit(500);
account.withdraw(300);
account.show(); // 山田 さんの残高: 1200 円

try {
  account.withdraw(5000);
} catch (error) {
  console.log("エラー:", error.message);
}
````
