# p5.js 入門チュートリアル

このノートブックでは、プログラミング初心者向けに **p5.js** を使った「絵を描く・動かす・操作する」プログラミングを学びます。
p5.js は、図形やアニメーションを簡単に描けるように作られた JavaScript のライブラリです。
すべてブラウザ上（JupyterLite の p5.js カーネル）で動くので、環境構築は不要です。

## このノートブックの使い方

- コードセルをクリックして **Shift + Enter** を押すと、そのセルが実行され、次のセルに移動します。
- **上から順番に、1 セルずつ** 実行してください。
- スケッチ（p5.js のプログラム）は、`setup` と `draw` という 2 つの関数を定義して作ります。
- `%show` と書いたセルを実行すると、その時点のスケッチが表示されます（`%show 100% 500px` のように表示サイズも指定できます）。
- 途中で分からなくなったら、メニューの **Kernel → Restart Kernel and Clear Outputs of All Cells...** でカーネルを再起動し、最初から実行し直せます。
- **Run All Cells** で一気に実行すると、すべての `%show` に最後のスケッチだけが表示されます（下の「カーネルの仕組み」を参照）。**1 セルずつ** 進めてください。

## p5.js の 2 つの関数

| 関数 | いつ呼ばれるか | 主な役割 |
|---|---|---|
| `setup()` | スケッチの開始時に **1 回だけ** | キャンバスを作る、初期設定をする |
| `draw()` | `setup()` の後、**1 秒間に約 60 回** ずっと繰り返し | 図形を描く（繰り返し描くことでアニメーションになる） |

## キャンバスの座標

- 原点 `(0, 0)` は **左上** です
- x は右に行くほど大きく、y は **下に行くほど大きく** なります（算数のグラフとは上下が逆です）
- 単位はピクセル（px）です

> **このカーネルの仕組み（重要）**: p5.js カーネルは「現在のスケッチ」を **1 つだけ** 持っています。コードセルを実行するたびに `setup` / `draw` などの定義が更新され、`%show` はその時点のスケッチを表示します。
> そのため、**新しいコードセルを実行すると、すでに表示されている `%show` の出力もすべて最新のスケッチに切り替わります**（前のスケッチの表示は残りません）。前のスケッチをもう一度見たいときは、そのコードセルを実行し直してから `%show` を実行してください。
> このノートブックでは、節ごとに `setup` と `draw` を定義し直して新しいスケッチを作っていきます。

JavaScript の文法そのものは `javascript-tutorial.ipynb` で学べます。ここでは「描く」ことに集中しましょう。

## 1. はじめてのスケッチ

まずは、キャンバスを作って円を 1 つ描いてみましょう。次のセルを実行してから、その下の `%show` セルを実行します。

```p5js
function setup() {
  createCanvas(400, 300);   // 幅 400px、高さ 300px のキャンバスを作る
}

function draw() {
  background(220);          // 背景を明るいグレー（0 = 黒 〜 255 = 白）で塗りつぶす
  circle(200, 150, 100);    // 中心 (200, 150)、直径 100 の円を描く
}
```

```p5js
%show
```

灰色の四角（キャンバス）の真ん中に白い円が表示されれば成功です。

- `createCanvas(幅, 高さ)` でキャンバスを作ります
- `background(色)` はキャンバス全体を塗りつぶします
- `circle(x, y, 直径)` は円を描きます。`(200, 150)` はキャンバス (400 × 300) のちょうど中央です

数値を変えて `%show` をもう一度実行してみましょう。たとえば `circle(100, 50, 30)` にすると、左上の小さな円になります。

## 2. 基本の図形

p5.js にはいろいろな図形を描く関数があります。

| 関数 | 描くもの | 引数 |
|---|---|---|
| `line(x1, y1, x2, y2)` | 線 | 始点と終点 |
| `rect(x, y, w, h)` | 四角形 | 左上の座標、幅、高さ |
| `square(x, y, s)` | 正方形 | 左上の座標、一辺の長さ |
| `circle(x, y, d)` | 円 | 中心の座標、直径 |
| `ellipse(x, y, w, h)` | 楕円 | 中心の座標、幅、高さ |
| `triangle(x1, y1, x2, y2, x3, y3)` | 三角形 | 3 つの頂点 |
| `arc(x, y, w, h, start, stop)` | 円弧 | 中心、幅、高さ、開始角と終了角（ラジアン） |
| `point(x, y)` | 点 | 座標 |

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(255);

  line(20, 20, 120, 80);                    // 線: (20, 20) から (120, 80) へ
  rect(150, 20, 100, 60);                   // 四角形: 左上 (150, 20)、幅 100、高さ 60
  square(280, 20, 60);                      // 正方形: 左上 (280, 20)、一辺 60
  circle(70, 160, 80);                      // 円: 中心 (70, 160)、直径 80
  ellipse(200, 160, 100, 60);               // 楕円: 中心 (200, 160)、幅 100、高さ 60
  triangle(280, 200, 340, 120, 390, 200);   // 三角形: 3 つの頂点
  arc(70, 260, 80, 60, 0, PI);              // 円弧: 0 〜 PI（半円）
  strokeWeight(8);                          // 線の太さを 8 に
  point(200, 260);                          // 点（太さ 8 なので見える）
  strokeWeight(1);                          // 元に戻す
}
```

```p5js
%show
```

> 角度は **ラジアン** で指定します。`PI` は 180 度、`HALF_PI` は 90 度、`TWO_PI` は 360 度です。`radians(90)` のように度からラジアンに変換することもできます。

## 3. 色

色の指定方法はいくつかあります。

| 書き方 | 例 | 意味 |
|---|---|---|
| 1 つの数値 | `fill(100)` | グレースケール（0 = 黒 〜 255 = 白） |
| 3 つの数値 | `fill(255, 0, 0)` | RGB（赤, 緑, 青 それぞれ 0 〜 255） |
| 4 つの数値 | `fill(255, 0, 0, 100)` | RGB + 透明度（0 = 透明 〜 255 = 不透明） |
| 文字列 | `fill("#4CAF50")`, `fill("orange")` | カラーコード、色の名前 |

- `fill(色)`: 図形の **塗りつぶし** の色
- `stroke(色)`: 図形の **線（輪郭）** の色
- `strokeWeight(太さ)`: 線の太さ
- `noFill()` / `noStroke()`: 塗りつぶし / 線 をなくす

これらの設定は、一度呼ぶと **その後に描く図形すべて** に適用されます。

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(30, 30, 60);        // RGB で背景色（濃い紺色）

  fill(255, 100, 100);           // 塗りの色（薄い赤）
  stroke(255);                   // 線の色（白）
  strokeWeight(4);               // 線の太さ
  circle(100, 100, 120);

  fill("#4CAF50");               // 16 進数のカラーコード（緑）
  noStroke();                    // 線を描かない
  rect(200, 50, 150, 100);

  fill("orange");                // 色の名前
  stroke(0);
  strokeWeight(1);
  triangle(60, 280, 140, 180, 220, 280);

  fill(100, 200, 255, 150);      // 4 つ目の値は透明度。重なった部分が透けて見える
  noStroke();
  circle(300, 220, 120);
  fill(255, 255, 100, 150);
  circle(340, 250, 120);
}
```

## 4. 動かす（アニメーション）

`draw()` は 1 秒間に約 60 回呼ばれます。**関数の外で変数を作り、`draw()` のたびに少しずつ値を変える** と、図形が動いて見えます。

```p5js
let x = 0;   // 円の x 座標。draw() の外で作るので、値が draw() をまたいで保存される

function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(220);         // 毎回背景を塗り直す（前のフレームの円を消す）
  circle(x, 150, 60);

  x = x + 2;               // 次のフレームでは 2px 右に描く
  if (x > width) {         // 右端を超えたら左端に戻す（width はキャンバスの幅）
    x = 0;
  }
}
```

```p5js
%show
```

試してみましょう:

- `x = x + 2` の `2` を大きくすると速くなります
- `background(220);` の行を削除（または `//` でコメントアウト）して `%show` すると、前のフレームが消えずに残り、軌跡が描かれます
- `width` と `height` はキャンバスの幅と高さが入っている特別な変数です。`frameCount` には開始からのフレーム数が入っています

## 5. 条件分岐で跳ね返らせる

速度を変数にして、端に当たったら符号を反転させると、ボールが跳ね返るようになります。

```p5js
let ballX = 200;   // ボールの位置
let ballY = 150;
let speedX = 3;    // 1 フレームあたりの移動量（速度）
let speedY = 2;

function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(20);
  fill(255, 200, 0);
  noStroke();
  circle(ballX, ballY, 40);

  ballX += speedX;   // 位置に速度を足す
  ballY += speedY;

  // 端に当たったら速度の向きを反転（半径 20 を考慮）
  if (ballX < 20 || ballX > width - 20) {
    speedX = -speedX;
  }
  if (ballY < 20 || ballY > height - 20) {
    speedY = -speedY;
  }
}
```

```p5js
%show
```

## 6. マウスで操作する

`mouseX` と `mouseY` には、キャンバス上のマウスの座標が常に入っています。`mouseIsPressed` はボタンが押されている間 `true` になります。

スケッチの上にマウスを動かしてみましょう。

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(240);

  if (mouseIsPressed) {
    fill(255, 80, 80);     // ボタンを押している間は赤
  } else {
    fill(80, 80, 255);     // 離しているときは青
  }
  noStroke();
  circle(mouseX, mouseY, 50);   // マウスの位置に円を描く

  // 座標を文字で表示
  fill(0);
  textSize(16);
  text("x: " + mouseX + "  y: " + mouseY, 10, 24);
}
```

```p5js
%show
```

## 7. イベント関数（クリック・キー入力）

「押された瞬間に 1 回だけ」処理をしたいときは、決まった名前の関数を定義します。p5.js が自動的に呼び出してくれます。

| 関数 | 呼ばれるタイミング |
|---|---|
| `mousePressed()` | マウスボタンが押されたとき |
| `mouseReleased()` | マウスボタンが離されたとき |
| `mouseDragged()` | ボタンを押しながらマウスを動かしたとき |
| `keyPressed()` | キーが押されたとき（押されたキーは `key` に入る） |

次のスケッチは簡単なお絵かきツールです。**ドラッグで線を描き、クリックで色付きの点を打ち、`c` キーで消去** します（キー入力は、先にスケッチをクリックして選択してから）。

```p5js
function setup() {
  createCanvas(400, 300);
  background(255);   // 背景は最初に 1 回だけ塗る（描いた線を残すため）
}

function draw() {
  // ドラッグしている間、前回のマウス位置 (pmouseX, pmouseY) から今の位置へ線を引く
  if (mouseIsPressed) {
    stroke(0);
    strokeWeight(4);
    line(pmouseX, pmouseY, mouseX, mouseY);
  }
}

// マウスボタンが押された瞬間に 1 回呼ばれる
function mousePressed() {
  fill(random(255), random(255), random(255));   // random(255) は 0 〜 255 のランダムな数
  noStroke();
  circle(mouseX, mouseY, 20);
}

// キーが押された瞬間に 1 回呼ばれる
function keyPressed() {
  if (key === "c") {
    background(255);   // c キーで全部消す
  }
}
```

```p5js
%show
```

> **ポイント**: このスケッチでは `draw()` の中で `background()` を呼んでいません。そのため描いた線が消えずに残ります。

## 8. 繰り返しで模様を作る

`for` 文を使うと、たくさんの図形を規則的に並べられます。二重ループにすると格子状になります。

`map(値, 元の最小, 元の最大, 変換後の最小, 変換後の最大)` は、値をある範囲から別の範囲に変換する便利な関数です。

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(255);
  noStroke();

  // 50px 間隔で円を格子状に並べる
  for (let y = 25; y < height; y += 50) {
    for (let x = 25; x < width; x += 50) {
      // 位置によって色を変える: x が大きいほど赤く、y が大きいほど青く
      let r = map(x, 0, width, 50, 255);
      let b = map(y, 0, height, 50, 255);
      fill(r, 100, b);
      circle(x, y, 40);
    }
  }
}
```

```p5js
%show
```

## 9. 乱数（random）とノイズ（noise）

- `random(max)` は 0 以上 max 未満のランダムな数、`random(min, max)` は min 以上 max 未満のランダムな数を返します
- `noise(t)` は 0 〜 1 の **なめらかに変化する** 乱数を返します。t を少しずつ増やすと、自然な揺れや動きが作れます

```p5js
function setup() {
  createCanvas(400, 300);
  frameRate(10);   // 1 秒間に 10 回だけ draw() を呼ぶ（ゆっくりにする）
}

function draw() {
  background(255, 30);   // 2 つ目の値は透明度。薄く塗るので前のフレームがうっすら残る

  // ランダムな位置・大きさ・色の円を 1 つ描く
  let x = random(width);
  let y = random(height);
  let d = random(10, 60);
  fill(random(255), random(255), random(255), 200);
  noStroke();
  circle(x, y, d);
}
```

```p5js
%show
```

```p5js
let t = 0;   // noise() に渡す「時間」

function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(230);

  // noise() は 0 〜 1 を返すので、キャンバスの幅・高さを掛けて座標にする
  let x = noise(t) * width;
  let y = noise(t + 100) * height;   // 別の値を渡すと別の動きになる

  fill(60, 120, 220);
  noStroke();
  circle(x, y, 40);

  t += 0.01;   // 小さく増やすほどゆっくり動く
}
```

```p5js
%show
```

`random()` の円はカクカクと現れるのに対し、`noise()` の円はふわふわと自然に動くのが分かります。

## 11. 座標系を動かす（translate / rotate / push / pop）

図形を回転させたいときは、図形ではなく **座標系** を動かします。

- `translate(x, y)`: 原点を (x, y) に移動する
- `rotate(角度)`: 座標系を原点を中心に回転する（ラジアン）
- `push()` / `pop()`: 座標系や色などの設定を保存 / 復元する

`push()` と `pop()` で囲むと、その中での変更が外に影響しません。

```p5js
let angle = 0;

function setup() {
  createCanvas(400, 300);
  rectMode(CENTER);   // rect の (x, y) を「左上」ではなく「中心」として扱う
}

function draw() {
  background(30);

  push();                            // ここから座標系を変更
  translate(width / 2, height / 2);  // 原点をキャンバスの中央に移動
  rotate(angle);                     // 座標系を回転
  fill(255, 150, 0);
  noStroke();
  rect(0, 0, 120, 120);              // 原点（= 中央）に四角形を描く
  pop();                             // 座標系を元に戻す

  // pop() の後なので、この円は回転の影響を受けない
  fill(255);
  circle(50, 50, 30);

  angle += 0.02;   // 毎フレーム少しずつ回転させる
}
```

```p5js
%show
```

## 12. 関数で整理する

同じ形を何度も描くときは、自分で関数を作ると便利です。引数で位置や大きさを変えられます。

`cos()` と `sin()` を使うと、円周上の点の座標を求められます（角度 a のとき `x + cos(a) * 半径`, `y + sin(a) * 半径`）。

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(200, 230, 255);
  drawFlower(100, 150, 30);
  drawFlower(220, 100, 50);
  drawFlower(320, 200, 40);
}

// 自分で作った関数: 中心 (x, y) に、大きさ size の花を描く
function drawFlower(x, y, size) {
  noStroke();
  fill(255, 120, 180);
  for (let i = 0; i < 6; i++) {
    let a = TWO_PI / 6 * i;    // 花びらの角度（360 度を 6 等分）
    circle(x + cos(a) * size, y + sin(a) * size, size);
  }
  fill(255, 220, 0);
  circle(x, y, size);          // 花の中心
}
```

```p5js
%show
```

## 13. 配列とクラスでたくさんのものを動かす

ボールをたくさん動かすには、「ボール 1 個分のデータと動き」をクラスにまとめ、配列に入れて管理します。

- `class Ball` に位置・速度・色と、動かす `update()`・描く `show()` を定義
- `balls` 配列にたくさんの `Ball` を入れ、`draw()` で 1 つずつ動かして描く
- クリックするとその場所にボールが増えます

```p5js
class Ball {
  constructor(x, y) {
    this.x = x;
    this.y = y;
    this.vx = random(-3, 3);   // x 方向の速度
    this.vy = random(-3, 3);   // y 方向の速度
    this.d = random(20, 50);   // 直径
    this.c = color(random(255), random(255), random(255));   // 色
  }

  update() {
    this.x += this.vx;
    this.y += this.vy;
    if (this.x < 0 || this.x > width) this.vx *= -1;    // 端で跳ね返る
    if (this.y < 0 || this.y > height) this.vy *= -1;
  }

  show() {
    noStroke();
    fill(this.c);
    circle(this.x, this.y, this.d);
  }
}

let balls = [];   // Ball をたくさん入れる配列

function setup() {
  createCanvas(400, 300);
  for (let i = 0; i < 10; i++) {
    balls.push(new Ball(random(width), random(height)));   // 最初に 10 個作る
  }
}

function draw() {
  background(20);
  for (const b of balls) {   // 配列の中のボールを 1 つずつ
    b.update();
    b.show();
  }
  fill(255);
  textSize(14);
  textAlign(LEFT, TOP);
  text("クリックでボールを追加: " + balls.length + " 個", 10, 10);
}

// クリックした場所にボールを追加（7 章の mousePressed を上書き）
function mousePressed() {
  balls.push(new Ball(mouseX, mouseY));
}
```

```p5js
%show
```

### 問題 2: 大きさが変わる円

キャンバスの中央で、円の直径がなめらかに大きくなったり小さくなったりを繰り返すスケッチを作ってください。

ヒント: `sin(frameCount * 0.05)` は -1 〜 1 の間をなめらかに往復します。`map()` で好きな範囲に変換できます。

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(30);
  // ここにコードを書いてください

}
```

```p5js
%show
```

**解答例**

```p5js
function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(30);

  let d = map(sin(frameCount * 0.05), -1, 1, 20, 200);   // -1〜1 を 20〜200 に変換
  fill(0, 200, 255);
  noStroke();
  circle(width / 2, height / 2, d);
}
```

```p5js
%show
```

### 問題 3: マウスを追いかける円

マウスの位置に **少し遅れてなめらかに** ついてくる円を作ってください。

ヒント: `lerp(現在の値, 目標の値, 0.1)` は、現在の値から目標の値に 10% だけ近づいた値を返します。毎フレーム呼ぶと、目標に向かってなめらかに近づきます。

```p5js
let followX = 200;   // 円の現在位置
let followY = 150;

function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(240);
  // ここにコードを書いてください（followX, followY を mouseX, mouseY に近づけて円を描く）

}
```

```p5js
%show
```

**解答例**

```p5js
let followX = 200;
let followY = 150;

function setup() {
  createCanvas(400, 300);
}

function draw() {
  background(240);

  followX = lerp(followX, mouseX, 0.1);   // マウスに 10% ずつ近づく
  followY = lerp(followY, mouseY, 0.1);

  fill(255, 100, 150);
  noStroke();
  circle(followX, followY, 50);
}
```

```p5js
%show
```
