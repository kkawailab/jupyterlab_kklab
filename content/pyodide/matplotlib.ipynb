## Matplotlib

```python
import piplite
await piplite.install("matplotlib-fontja==1.1.0")

import matplotlib_fontja
import numpy as np
import matplotlib.pyplot as plt

matplotlib_fontja.japanize()
```

```python
x = np.linspace(0, 10, 1000)
plt.plot(x, np.sin(x))
plt.title("日本語フォントの確認")
plt.xlabel("時間")
plt.ylabel("振幅");
```

```python
plt.show()
```

## Matplotlib: reusable plotting APIs

```python
plt.close("all")
```

```python
x = np.linspace(0, 10, 1000)
figure, axis = plt.subplots()
axis.plot(x, np.cos(x))
axis.set_title("コサイン波");
```
